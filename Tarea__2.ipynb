se crean 2 listas
se introduce la funcion zip para producir un iterator (para tomar los valores de estas listas y unirlos en una tupla)
se crea el diccionario students y se le asignan los keys&values del zipstudebts

```python
##Dict Comprehensions
#Let’s say we have two lists that we want to combine into a
#dictionary, like a list of students and a list of their heights, in inches:

names = ['Jenny', 'Alexus', 'Sam', 'Grace']
heights = [61, 70, 67, 64]
  #Python allows you to create a dictionary using
  #a dict comprehension, with this syntax:

zipStudents = zip(names, heights)
print("zipStudents: ", zipStudents)

students = {key:value for key, value in zip(names, heights)}
#students is now {'Jenny': 61, 'Alexus': 70, 'Sam': 67, 'Grace': 64}
print(students)

#zip() combines two lists into an iterator of tuples with the list elements paired together. This dict comprehension:
```

# Tuples

Corresponde a una estructura similar a las listas, la diferencia está
en que no se pueden modificar una vez creadas, es decir que son inmutables:

se convierte la lista my_lista en una tupla utilizando tuple()
la nueva tupla se guarda en my_tupla

```python
#Convertir una lista a tupla:prin
print("###########################")
print("###########################")
print("###########################")
print("############TUPLAS#########")
my_tupla = tuple(my_lista)
print()
print()
print("my_tuple: ", my_tupla)
```

se accede a los elementos de la tupla utilizando su posicion

```python
print(my_tupla[0])
print(my_tupla[2])
```

se verifica si Rojo se encuentra dentro de la tupla
se utiliza in y devuelve un valor booleano
tambien se utiliza count() para saber cuantas veces aparece Rojo


```python
#Evaluar si un elemento está contenido en la tupla (Devuelve un valor booleano)
print('Rojo' in my_tupla)
print(my_tupla.count('Rojo'))
```

se crea una tupla con un solo elemento

```python
#Tupla con un solo elemento
my_tupla_unitaria = ('Blanco')
print(my_tupla_unitaria)
```

se crea una tupla sin utilizar parentesis
los elementos se separan utilizando comas

```python
#Empaquetado de tupla, tupla sin paréntesis
my_tupla = 'Gaspar', 5, 8, 1999
print(my_tupla)
```

se desempaquetan los valores de la tupla
cada valor se guarda en una variable siguiendo el mismo orden


```python
#Desempaquetado de tupla, se guardan los valores en orden de las variables
nombre, dia, mes, año = my_tupla
print(nombre)
print(dia)
print(mes)
print(año)
```
