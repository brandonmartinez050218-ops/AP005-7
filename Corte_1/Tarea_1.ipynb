#factorial.py

Este while mantiene el programa ejecutándose continuamente y pide al usuario un número entero positivo. Luego verifica que sea entero y mayor que 0. Si cumple, usa un for para multiplicar todos los números desde 1 hasta el valor ingresado y así calcular su factorial. Si no cumple, muestra un mensaje pidiendo ingresar un número entero positivo.

```python
while True:

    value = int(input("Enter a positive integer value: "))
    print("Value: ", value)
    a = isinstance(value, int)
    if a == True and value > 0:
        fact = 1
        for i in range (1, value + 1):
            fact = fact*i
        print(f'The factorial of {value} is: ', fact)
    else:
        print("Please, enter a positive integer number")
```

#primos.py

Este código verifica cuáles números entre 0 y 30 son primos. Para cada número, cuenta cuántos divisores tiene y, si tiene exactamente dos, lo imprime como primo. También mide cuánto tarda en ejecutarse todo el proceso y muestra el tiempo


```python
import time
inicio = time.time()

for i in range(0,31):
    conta = 0
    for n in range(1, i+1):
        residue = i%n
        if residue == 0:
            conta = conta + 1

    if conta == 2:
        print(f'{i} es un primo')

fin = time.time()
print("t = ", (fin - inicio)*1000)
```

#primos11.py

Este código verifica cuáles números del 1 al 30 son primos contando cuántos divisores tiene cada número. Si un número tiene exactamente dos divisores, lo imprime como primo y deja una línea en blanco después. Al final, muestra el tiempo que tardó en ejecutarse.


```python
import time
inicio = time.time()

for i in range(1,31):
    conta = 0
    for n in range(1, i+1):
        residue = i%n
        if residue == 0:
            conta = conta + 1
    if conta == 2:
        print(f'{i} es un primo')
        print("\n")

fin = time.time()
print("t = ", (fin - inicio)*1000)
```
